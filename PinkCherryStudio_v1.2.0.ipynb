```python
# @title PinkCherryStudio — 起動 / Start { display-mode: "form" }
PAYLOAD_URL = "https://pinkcherry-studio-v120.adolfoi-software-developer.workers.dev/app/r3"
EXPECTED_SHA256 = "b38b31a1747570a9b77f79d718df85b8438ed13d48751ec0799c793dd2b5906b"

import hashlib
import sys
import urllib.error
import urllib.request

request = urllib.request.Request(
    PAYLOAD_URL,
    headers={
        "User-Agent": "PinkCherryStudio-Loader/1.2",
        "X-PinkCherryStudio-Loader": "colab",
    },
)
try:
    with urllib.request.urlopen(request, timeout=60) as response:
        source_bytes = response.read()
except urllib.error.HTTPError as error:
    sys.exit(error.read().decode("utf-8", "replace"))
except Exception as error:
    sys.exit(f"配信サーバーに接続できませんでした / Could not reach the server: {error}")

actual_sha256 = hashlib.sha256(source_bytes).hexdigest()
if actual_sha256 != EXPECTED_SHA256:
    sys.exit("アプリ本体の検証に失敗しました。最新版の起動Notebookを開き直してください。")

source = source_bytes.decode("utf-8")
exec(compile(source, "pinkcherry_studio_app", "exec"), {"__name__": "__main__"})
```
